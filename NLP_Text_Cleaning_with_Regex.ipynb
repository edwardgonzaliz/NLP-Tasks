```python
text = """This is some sample data for practicing text processing exercises.
The quick brown fox jumps over the lazy dog, running and playing.
My email is user@example.com and another one is support.team@sub.domain.co.uk.
Please call me at 123-456-7890 or +1 (555) 123-4567. My other number is 07911 123456.
The meeting is scheduled for 2023-10-27 and also on 11/15/2024. Today's date is October 26, 2023.
We are using a computer system for data analysis. The old computer was slow. This is important data.
Here are some numbers: 123, 45.67, and 1,000. This is 1.25. Version 2.0 is out.
#Python #Regex #DataScience. I am enjoying learning these skills. #AI #NLP.
Visit our website at http://www.example.com or https://anothersite.org.
He said, \"Hello world!\" How are you?
This is a test sentence with various punctuation!.,;:?"
"""
```

```python
import re
```

```python
matches = re.findall("data", text)
print(matches)
```

```python
numbers = re.findall("\d", text)
print(numbers)
```

```python
numbers = re.findall("\d+", text)
print(numbers)
```

```python
new_text = re.sub("computer", "system", text)
print(new_text)
```

```python
new_corpus_space = re.split(" ",text)
print(new_corpus_space)
```

```python
new_corpus = re.split("\n", text)
print(new_corpus)
```

```python
text1 = """This is some sample data for practicing text processing exercises.
The quick brown fox jumps over the lazy dog, running and playing.
My email is user@example.com and another one is support.team@sub.domain.co.uk.
Please call me at 123-456-7890 or +1 (555) 123-4567. My other number is 07911 123456.
The meeting is scheduled for 2023-10-27 and also on 11/15/2024. Today's date is October 26, 2023.
We are using a computer system for data analysis. The old computer was slow. This is important data.
Here are some numbers: 123, 45.67, and 1,000. This is 1.25. Version 2.0 is out.
#Python #Regex #DataScience. I am enjoying learning these skills. #AI #NLP.
Visit our website at http://www.example.com or https://anothersite.org.
He said, \"Hello world!\" How are you?
This is a test sentence with various punctuation!.,;:?"
"""
```

```python
He_start = re.findall("^He", text1)
print(He_start)
```

```python
email = re.findall(r'\S+@\S+', text)
print(email)
```

```python
hashtags = re.findall(r'#\S+', text)
print(hashtags)
```

```python
clean_text = re.sub(r'[^\w\s]', '', text)
print(clean_text)
```

```python
ing_end = re.findall(r'\b\w+ing\b', text)
print(ing_end)
```

```python
k = input('Enter Password:')
if re.fullmatch(r'[A-Za-z0-9@#$%^&+=]{8,}', k):
    print("Valid Password")
else:
    print("Invalid Password")
```
